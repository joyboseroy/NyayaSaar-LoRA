```python
df = pd.DataFrame(examples)

dataset_small = Dataset.from_pandas(df)

print(dataset_small)
```

```python
# ============================================================
# Prompt Formatting
# ============================================================

def formatting_func(example):
    text = f"""
### Instruction:
Simplify the following Indian legal reasoning into plain English.

### Legal Text:
{example['input']}

### Plain English:
{example['output']}
"""
    return {"text": text}

dataset_small = dataset_small.map(formatting_func)

print(dataset_small[0]["text"])
```

```python
# ============================================================
# Load Tiny Model
# ============================================================

max_seq_length = 2048

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = "unsloth/Qwen2.5-0.5B-Instruct-bnb-4bit",
    max_seq_length = max_seq_length,
    dtype = None,
    load_in_4bit = True,
)
```

```python
# ============================================================
# Add LoRA Adapters
# ============================================================

model = FastLanguageModel.get_peft_model(
    model,
    r = 16,
    target_modules = [
        "q_proj",
        "k_proj",
        "v_proj",
        "o_proj",
        "gate_proj",
        "up_proj",
        "down_proj",
    ],
    lora_alpha = 16,
    lora_dropout = 0,
    bias = "none",
    use_gradient_checkpointing = "unsloth",
    random_state = 3407,
)
```

```python
# ============================================================
# Trainer
# ============================================================

trainer = SFTTrainer(
    model = model,
    tokenizer = tokenizer,
    train_dataset = dataset_small,
    dataset_text_field = "text",
    max_seq_length = max_seq_length,

    args = TrainingArguments(
        per_device_train_batch_size = 2,
        gradient_accumulation_steps = 4,
        warmup_steps = 5,
        max_steps = 30,
        learning_rate = 2e-4,
        logging_steps = 1,
        optim = "adamw_8bit",
        weight_decay = 0.01,
        lr_scheduler_type = "linear",
        seed = 3407,
        output_dir = "outputs",
    ),
)
```

```python
# ============================================================
# Train
# ============================================================

trainer.train()
```

```python
# ============================================================
# Inference
# ============================================================

FastLanguageModel.for_inference(model)

test_prompt = """
### Instruction:
Simplify the following Indian legal reasoning into plain English.

### Legal Text:
Issue: Whether the writ petition is maintainable.

Rule: Writ jurisdiction applies when legal rights are violated.

Application: The petitioner argued authorities acted arbitrarily.

Conclusion: Petition allowed.

### Plain English:
"""

inputs = tokenizer(
    [test_prompt],
    return_tensors="pt"
).to("cuda")

outputs = model.generate(
    **inputs,
    max_new_tokens = 200,
    use_cache = True,
)

response = tokenizer.batch_decode(outputs)

print(response[0])
```

```python
# ============================================================
# Readability Evaluation
# ============================================================

legal_text = """
The petitioner herein seeks issuance of a writ of mandamus.
"""

plain_text = """
The petitioner asked the court to issue an order.
"""

print("Legal Text Grade Level:")
print(textstat.flesch_kincaid_grade(legal_text))

print("\nPlain English Grade Level:")
print(textstat.flesch_kincaid_grade(plain_text))
```

```python
# ============================================================
# Save LoRA
# ============================================================

model.save_pretrained("nyayasaar_lora")
tokenizer.save_pretrained("nyayasaar_lora")

print("LoRA model saved!")
```
